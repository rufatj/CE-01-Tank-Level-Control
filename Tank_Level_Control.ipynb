```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

A = float(input("Tank area (m²): "))
Q_in =float(input("Inlet flow (m³/s): "))
Q_out =float(input("Outlet flow (m³/s): "))

def tank_model(t,h):
  dhdt=(Q_in - Q_out)/A
  return dhdt

time = 20      # simulation time (seconds)
dt = 1         # time step (seconds)
h = 0          #  initial water level (m)

times=[]
levels=[]

for t in range(time + 1):

    dhdt =(Q_in - Q_out) /A

    h =h + dhdt*dt

    if h<0:
        h = 0

    times.append(t)
    levels.append(h)



plt.figure(figsize=(8,5))
plt.plot(times,levels)
2
plt.xlabel("Time (s)")
plt.ylabel("Water Level (m)")
plt.title("Tank Level Simulation")


plt.grid(True)  #arxa fondaki xetler
plt.show()
```
